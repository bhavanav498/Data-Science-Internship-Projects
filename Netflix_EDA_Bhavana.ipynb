```python
import pandas as pd

df = pd.read_csv("netflix_titles.csv")
df.head()
```

```python
df[df['director'].isnull()].head()
```

```python
df[df['country'].isnull()].head()
```

```python
df['director'].fillna('Unknown', inplace=True)
df['country'].fillna('Unknown', inplace=True)
df['cast'].fillna('Unknown', inplace=True)
```

```python
df.isnull().sum()
```

```python
df['date_added'].fillna(df['date_added'].mode()[0], inplace=True)
df['rating'].fillna(df['rating'].mode()[0], inplace=True)
df['duration'].fillna(df['duration'].mode()[0], inplace=True)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.to_csv("netflix_cleaned.csv", index=False)
```

```python
import matplotlib.pyplot as plt

df['type'].value_counts().plot(kind='bar',color = ['purple', 'orange'])
plt.title('Movies vs TV Shows on Netflix')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()
```

1. Movies vs TV Shows

Movies constitute a significantly larger portion of Netflix's catalog compared to TV Shows.


```python
df['country'].value_counts().head(10).plot(kind='bar',color='teal')
plt.title('Top 10 Content Producing Countries')
plt.xlabel('Country')
plt.ylabel('Number of Titles')
plt.show()
```

2. Top 10 Countries Producing Content


The United States contributes the highest number of titles on Netflix, followed by a few other major content-producing countries.

```python
df['release_year'].value_counts().sort_index().plot(figsize=(12,5))
plt.title('Netflix Content Released by Year')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.show()
```

3. Content Released by Year

Netflix content releases increased rapidly after 2015, indicating substantial growth in content production and acquisition.

```python
df['rating'].value_counts().plot(kind='bar')
plt.title('Distribution of Ratings')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```

4. Ratings Distribution

TV-MA is the most common content rating, suggesting a strong focus on mature audiences.

```python
genres = df['listed_in'].str.split(', ', expand=True).stack()

genres.value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Genres on Netflix')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.show()
```

5. Top Genres

International Movies and Drama-related genres are among the most popular categories available on Netflix.

```python
import seaborn as sns
import matplotlib.pyplot as plt

numeric_df = df[['release_year']]

corr = numeric_df.corr()

sns.heatmap(corr, annot=True)
plt.title("Correlation Matrix")
plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df_corr = df.copy()

df_corr['type'] = le.fit_transform(df_corr['type'])
df_corr['rating'] = le.fit_transform(df_corr['rating'])
df_corr['country'] = le.fit_transform(df_corr['country'])
```

```python
corr_matrix = df_corr[['type','rating','country','release_year']].corr()

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```
